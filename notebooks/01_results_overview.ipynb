```python
import json
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA = ROOT / 'results' / 'data'

def load(name):
    with open(DATA / name) as handle:
        return json.load(handle)
```

```python
models = {
    'random_mask_1x': load('random_mask_execution.json'),
    'random_mask_2x': load('random_mask_compute_matched_execution.json'),
    'joint_ce': load('joint_ce_execution.json'),
    'consistency': load('climbmix_execution.json'),
}
for name, result in models.items():
    print(f"{name:18s} agreement={100*result['argmax_agreement']:.2f}%  KL={result['symmetric_kl']:.3f}")
```

## Interpretation

Random-mask and joint-CE baselines preserve average language-model quality but do not produce functionally equivalent graphs. The consistency objective primarily improves agreement and KL rather than BPB.
